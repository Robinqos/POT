### 6B) `[Flags]`

**Úloha:**  
Definujte `[Flags]` enum `TextStyle` (None, Bold, Italic, Underline, Monospace).  
Implementujte metódu `Apply(string text, TextStyle style)`, ktorá obalí text značkami:  
Bold ⇒ `[B]text[/B]`, Italic ⇒ `[I]text[/I]`, Underline ⇒ `[U]text[/U]`, Monospace ⇒ `[M]text[/M]`.

**Čo máte urobiť:** definovať enum, doplniť `Apply`, **overiť** self-check.

```polyglot-notebook
// TODO: definujte enum TextStyle s príznakmi Bold, Italic, Monospace, Underline

string Apply(string text, TextStyle style)
{
    string result = text;
    if ((style & TextStyle.Monospace) == TextStyle.Monospace) result = $"[M]{result}[/M]";
    if ((style & TextStyle.Underline) == TextStyle.Underline) result = $"[U]{result}[/U]";

    return result;
}

// self-check
var s = Apply("CSharp", TextStyle.Bold | TextStyle.Italic);
Console.WriteLine(s); // [B][I]CSharp[/I][/B]
var t = Apply("code", TextStyle.Monospace | TextStyle.Underline);
Console.WriteLine(t); // [U][M]code[/M][/U]
```

## 9) Porovnajte správanie pri priradení pre class a struct.


```polyglot-notebook
public class RefPoint
{
    public int X { get; set; }
    public int Y { get; set; }
    public RefPoint(int x, int y) { X = x; Y = y; }
}

public struct ValuePoint
{
    public int X { get; set; }
    public int Y { get; set; }
    public ValuePoint(int x, int y) { X = x; Y = y; }
}

var r1 = new RefPoint(1, 1);
var r2 = r1;
r1.X *= 10;
Console.WriteLine($"RefPoint: r1.X={r1.X}, r2.X={r2.X}"); // RefPoint: r1.X=10, r2.X=10

var v1 = new ValuePoint(1, 1);
var v2 = v1;
v1.X *= 10;
Console.WriteLine($"ValuePoint: v1.X={v1.X}, v2.X={v2.X}"); // ValuePoint: v1.X=10, v2.X=1
```

## 10) `readonly struct` a `record struct` – porovnanie

### 10A) `readonly struct` Vector2:  
`readonly struct Vector2(double X, double Y)` s `Length()`.



```polyglot-notebook
public readonly struct Vector2
{
    public double X { get; }
    public double Y { get; }
    public Vector2(double x, double y) { X = x; Y = y; }
    public double Length() => Math.Sqrt(X*X + Y*Y);
}

var v1 = new Vector2(3, 4);
var v2 = v1;
Console.WriteLine(v1.Length()); // 5
Console.WriteLine(v2.Length()); // 5

var hs = new HashSet<Vector2>();
hs.Add(new Vector2(1,2));
hs.Add(new Vector2(1,2));
Console.WriteLine(hs.Count); // 1
```

## 11) Primárny konštruktor (`C# 12`) – bankový účet

**Úloha:**  
- Pridajte primárny konštruktor do triedy `class BankAccount` a vlastnosti `AccountId`, `Owner` iba na čítanie, `Balance` s `private set`. 

```polyglot-notebook
public class BankAccount
{


    public void Deposit(decimal amount)
    {
        if (amount <= 0) throw new ArgumentOutOfRangeException(nameof(amount));
        Balance += amount;
    }

    public void Withdraw(decimal amount)
    {
        if (amount <= 0) throw new ArgumentOutOfRangeException(nameof(amount));
        if (Balance < amount) throw new InvalidOperationException("Insufficient funds");
        Balance -= amount;
    }

    public override string ToString() => $"{AccountId} – {Owner}: {Balance:C}";
}

var acc = new BankAccount("SK00-123", "Ján");
acc.Deposit(100);
acc.Withdraw(40);
Console.WriteLine(acc); // SK00-123 – Ján: 60,00 €
```
